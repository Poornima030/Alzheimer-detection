## Interpreting the ablation

Per the project's research framing, the goal is **not** "did we beat Tripathy's reported
accuracy" — with so few images per class, small accuracy deltas between experiments may not be
statistically meaningful. Instead, look at:

- Does each addition (CBAM, Transformer, SSL) move F1/recall on the **minority classes**
  (MildDemented, ModerateDemented) specifically, where the baseline is most likely to struggle?
- Does SSL pretraining reduce the gap between train and validation accuracy (i.e. help
  generalization/data efficiency), even if final test accuracy is similar?
- Do the Grad-CAM/SHAP/LIME explanations from notebook 07 become more anatomically coherent
  as components are added, independent of the raw accuracy number?

Write these findings up as the actual contribution, with the numbers in this table as evidence
either way — including negative results (e.g. "Transformer branch overfits on OASIS's ~400
training images and does not outperform CBAM alone") are legitimate and publishable findings for
a thesis, not something to hide.
