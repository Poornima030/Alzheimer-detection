```python
# Block 9: visualize a few MRI samples (only works once images are in place)
import matplotlib.pyplot as plt
from src.preprocessing import load_and_preprocess_image
import tensorflow as tf

if len(train_paths) > 0:
    fig, axes = plt.subplots(1, min(5, len(train_paths)), figsize=(15, 3))
    if min(5, len(train_paths)) == 1:
        axes = [axes]
    for i, ax in enumerate(axes):
        img = load_and_preprocess_image(tf.constant(train_paths[i]))
        ax.imshow(img.numpy().squeeze(), cmap='gray')
        ax.set_title(cfg.CLASS_NAMES[train_labels[i]], fontsize=9)
        ax.axis('off')
    plt.tight_layout()
    plt.show()
else:
    print("No images to visualize yet — place images under data/images/ first.")
```
